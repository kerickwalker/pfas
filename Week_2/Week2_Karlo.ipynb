```python
import cv2
import numpy as np

from matplotlib import pyplot as plt
```

```python
# import video
cap = cv2.VideoCapture('Robots.mp4')
```

```python
# video test
ret, frame = cap.read()
while ret:
    cv2.imshow('image', frame)
    cv2.waitKey(20)
    ret,frame = cap.read()
cap.release()
cv2.destroyAllWindows()
```

```python
# Sparse Optical Flow
cap = cv2.VideoCapture('Robots.mp4')
ret, frame = cap.read()
while ret:
    img1 = frame.copy()
    
    ret,frame = cap.read()
    
    img2 = frame.copy()
    
    gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
    gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
    
    feat1 = cv2.goodFeaturesToTrack(gray1, maxCorners=1000, qualityLevel=0.2, minDistance = 7)
    feat2, status, error = cv2.calcOpticalFlowPyrLK(gray1, gray2, feat1, None)

    for i in range(len(feat1)):
        f10=int(feat1[i][0][0])
        f11=int(feat1[i][0][1])
        f20=int(feat2[i][0][0])
        f21=int(feat2[i][0][1])
        if (f10-f20)**2+(f21-f11)**2>5:    # filter out movement that is big enough (euclidian distance)
            cv2.line(img2, (f10,f11), (f20, f21), (0, 255, 0), 2)
            cv2.circle(img2, (f10, f11), 5, (0, 255, 0), -1)

    cv2.imshow('robots', img2)
    cv2.waitKey(20)

    ret, frame = cap.read()
cv2.waitKey(100)
cap.release()
cv2.destroyAllWindows()
```

```python
# Gunnar Farneback (Dense) Optical Flow
cap = cv2.VideoCapture('Robots.mp4')
ret, frame = cap.read()
(height, width, _) = frame.shape
while ret:
    img1 = frame.copy() # b, g, r
    
    ret,frame = cap.read()
    
    img2 = frame.copy() # b, g, r
    
    gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
    gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
    
    flow = cv2.calcOpticalFlowFarneback(gray1, gray2, None, 0.5, 3, 17, 3, 5, 1.5, 0)
    mag, ang = cv2.cartToPolar(flow[:,:, 0], flow[:,:, 1])

    empty_hsv = np.zeros((height, width,3), np.uint8)
    empty_hsv[:,:, 1] = 255  # high saturation
    
    mag[mag < 0.2 * mag.max()] = 0 
    ang[mag == 0] = 0

    empty_hsv[:,:, 0] = ang * 180 / np.pi / 2
    empty_hsv[:,:, 2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)
    rgb_rep = cv2.cvtColor(empty_hsv, cv2.COLOR_HSV2BGR)

    result = cv2.bitwise_or(rgb_rep, img2, mask=None)
    cv2.imshow('robots', result)
    cv2.waitKey(20)

    ret, frame = cap.read()
cv2.waitKey(100)
cap.release()
cv2.destroyAllWindows()
```

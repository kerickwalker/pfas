```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from math import gcd
```

```python
#Loading the video
cap = cv2.VideoCapture('Robots.mp4')    
```

Method 1 : Sparse Optical Flow 
This method consists in detecting and tracking features between each frames of the video. We are using OpenCV functions to do so.

```python
#Reading each image 
ret1, frame1 = cap.read()
while ret1:
    b1,g1,r1 = cv2.split(frame1) 
    img1 = cv2.merge([r1,g1,b1])
    gray1 = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
    #Detect the feature using Shi-Tomasi 
    feat1 = cv2.goodFeaturesToTrack(gray1, maxCorners = 150, qualityLevel = 0.02,
                minDistance = 20) 
    #Proceed the same with the next frame
    ret2,frame2 = cap.read()
    if ret2:
        b2,g2,r2 = cv2.split(frame2) 
        img2 = cv2.merge([r2,g2,b2])
        gray2 = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)
        #Track the features in the next frame
        feat2, status, error = cv2.calcOpticalFlowPyrLK(gray1, gray2, feat1, None)
    
        #Test the tracking by displaying the features movement on the latest frame
        for i in range(len(feat1)):
            f10=int(feat1[i][0][0])
            f11=int(feat1[i][0][1])
            f20=int(feat2[i][0][0])
            f21=int(feat2[i][0][1])
            dist = ((f10 -f20)**2 +(f11-f21)**2)**0.5
            #Put a threshold on how much of distance do we consider a displacement
            if dist >5:
                cv2.line(frame2, (f10,f11), (f20, f21), (0, 255, 0), 2)
                cv2.circle(frame2, (f10, f11), 5, (0, 255, 0), -1)
        ret1 = ret2
        frame1 = frame2
        cv2.imshow("image", frame2)
        cv2.waitKey(50)
cv2.destroyAllWindows()
cap.release()
```
